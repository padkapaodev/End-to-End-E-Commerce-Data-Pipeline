```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from delta.tables import *
```

```python
%sql
create or replace table project_cat.default.config_table(
    table_name string,
    file_path string,
    keys array<string>,
    schema_details map<string,string>,
    schema_location string,
    checkpoint_location string
)
```

```python
def upsert_into (df, keys:list) ->None:
    delta_obj = DeltaTable.forName(spark, "project_cat.default.config_table")
    (delta_obj.alias("t").merge(
        df.alias("s"),
        " AND ".join([f"t.{key} = s.{key}" for key in keys]))
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute())
```

```python
#dim_customers

data = [{
    "table_name": "dim_customers",
    "file_path": "project_cat.bronze_layer.dim_customers",
    "keys": ["customer_id"],
    "schema_details": {"customer_id": "string",
                        "customer_name": "string",
                        "customer_age": "integer",
                        "gender": "string",
                        "customer_segment": "string",
                        "customer_city":"string",
                        "customer_country":"string",
                        "region":"string",
                        "customer_postal_code":"integer",
                        "customer_acquisition_cost":"decimal(10,2)"},
    "schema_location": "/Volumes/project_cat/default/schemalocation/dim_customers/",
    "checkpoint_location": "/Volumes/project_cat/default/checkpoint/dim_customers/"
        }] 

schema = StructType([
    StructField("table_name", StringType()),
    StructField("file_path", StringType()),
    StructField("keys", ArrayType(StringType())),
    StructField("schema_details", MapType(StringType(), StringType())),
    StructField("schema_location", StringType()),
    StructField("checkpoint_location", StringType())
    ])

df = spark.createDataFrame(data, schema=schema)
upsert_into(df, ["table_name"])
```
